### Variables Description

### Variable Description

**1. Titles Dataset**

The Titles dataset consists of 15 variables that provide information about movies and television shows.

- **id:** A unique identification number assigned to each movie or television show.
- **title:** Represents the name of the movie or television show.
- **type:** Specifies whether the content is a movie or a TV show.
- **description:** Provides a brief summary or storyline of the content.
- **release_year:** Indicates the year in which the content was released.
- **age_certification:** Specifies the age rating or content classification.
- **runtime:** Represents the duration of the movie or show.
- **genres:** Describes the category or type of content, such as drama, comedy, action or thriller.
- **production_countries:** Indicates the countries associated with the production of the content.
- **seasons:** Represents the total number of seasons available for a television show.
- **imdb_id:** Contains the unique identification number assigned by IMDb.
- **imdb_score:** Represents the IMDb rating given to the content.
- **imdb_votes:** Indicates the total number of votes received on IMDb.
- **tmdb_popularity:** Represents the popularity score provided by The Movie Database (TMDB).
- **tmdb_score:** Represents the rating score provided by TMDB.

**2. Credits Dataset**

The Credits dataset consists of 5 variables that provide information about the actors and directors associated with each title.

- **person_id:** A unique identification number assigned to each person.
- **id:** Represents the identification number of the movie or TV show associated with the person.
- **name:** Contains the name of the actor or director.
- **character:** Specifies the character played by an actor in a movie or television show. This field may be empty for some records.
- **role:** Indicates the person's contribution to the title, such as ACTOR or DIRECTOR.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
df_titles_clean = df_titles.copy()
df_credits_clean = df_credits.copy()

df_titles_clean.drop_duplicates(inplace=True)
df_credits_clean.drop_duplicates(inplace=True)

for col in df_titles_clean.select_dtypes(include='object').columns:
    df_titles_clean[col] = df_titles_clean[col].str.strip()

for col in df_credits_clean.select_dtypes(include='object').columns:
    df_credits_clean[col] = df_credits_clean[col].str.strip()

numeric_columns = [
    'release_year', 'runtime', 'seasons',
    'imdb_score', 'imdb_votes',
    'tmdb_popularity', 'tmdb_score'
]

for col in numeric_columns:
    df_titles_clean[col] = pd.to_numeric(
        df_titles_clean[col], errors='coerce'
    )

print("Missing Values in Titles Dataset:")
print(df_titles_clean.isnull().sum())

print("\nMissing Values in Credits Dataset:")
print(df_credits_clean.isnull().sum())

print("\nDuplicates in Titles:",
      df_titles_clean.duplicated().sum())

print("Duplicates in Credits:",
      df_credits_clean.duplicated().sum())

print("\nCleaned Titles Shape:", df_titles_clean.shape)
print("Cleaned Credits Shape:", df_credits_clean.shape)

display(df_titles_clean.head())
display(df_credits_clean.head())
````

#### Chart - 4

````python
import ast

def extract_genres(value):
    if pd.isna(value):
        return []
    try:
        return ast.literal_eval(value)
    except (ValueError, SyntaxError):
        return [value]

genre_data = df_titles['genres'].apply(extract_genres)
all_genres = genre_data.explode()

genre_counts = all_genres.value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=genre_counts.values,
    y=genre_counts.index,
    palette='viridis'
)

plt.title('Top 10 Most Common Genres on Amazon Prime Video')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')

plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A horizontal bar chart was selected to identify and compare the most common genres available on Amazon Prime Video. Since a title can belong to multiple genres, analysing genre frequencies helps understand the variety of content represented in the dataset. This chart makes it easier to identify the most frequently occurring genres and compare their representation.

##### 2. What is/are the insight(s) found from the chart?

The chart shows that some genres are represented more frequently than others in the Amazon Prime Video catalogue. The most frequently occurring genres have higher counts, while other genres have comparatively fewer titles. This indicates that the content catalogue has different levels of representation across genres. The results also help identify popular content categories and areas where genre representation is relatively limited.

#### Chart - 14 - Correlation Heatmap

````python
numeric_data = df_titles[
    ['release_year', 'runtime', 'seasons',
     'imdb_score', 'imdb_votes',
     'tmdb_popularity', 'tmdb_score']
]

correlation_matrix = numeric_data.corr()

plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Numerical Variables', fontsize=14)
plt.tight_layout()
plt.show()
````

Stronger positive or negative correlations indicate variables that move together, while values closer to zero indicate weaker linear relationships. The chart also helps identify which variables have limited relationships with one another.

#### Chart - 15 - Pair Plot

````python
pairplot_cols = [
    'runtime',
    'imdb_score',
    'imdb_votes',
    'tmdb_popularity',
    'tmdb_score',
    'type'
]

pairplot_data = df_titles[pairplot_cols].dropna().copy()

if len(pairplot_data) > 1000:
    pairplot_data = pairplot_data.sample(1000, random_state=42)

sns.pairplot(
    pairplot_data,
    vars=pairplot_cols[:-1],
    hue='type',
    palette='deep',
    corner=True,
    plot_kws={'alpha': 0.45, 's': 18}
)

plt.suptitle('Pair Plot of Selected Numerical Variables', y=1.02)
plt.show()
````

##### 1. Why did you pick the specific chart?

A pair plot was selected because it displays the distributions of numerical variables along the diagonal and pairwise scatter plots outside the diagonal. It helps examine relationships among runtime, IMDb score, IMDb votes, TMDB popularity and TMDB score. The content type is used to distinguish movies and TV shows.

##### 2. What is/are the insight(s) found from the chart?

The pair plot provides an overview of the distributions and relationships among the selected numerical variables. It helps identify possible correlations, clusters and differences between movies and TV shows. These patterns can be examined alongside the correlation heatmap to understand how the numerical features are related.
